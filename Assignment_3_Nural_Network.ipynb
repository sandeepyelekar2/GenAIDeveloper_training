```python
#draw and visualize the Neural Network.
import matplotlib.pyplot as plt
# Helps us work with numerical data.
import numpy as np


```

```python

#features used by the Network
input_features = [
    "TransactionAmount",
    "TransactionTime",
    "MerchantCategory",
    "CustomerAge",
    "AccountBalance",
    "NumberOfTransactionsToday"
]

# 0 = Genuine transaction
# 1 = Fraudulent transaction
output_feature = "Fraud"


```

```python
# Number of neurons in each layer
input_neurons = 6
hidden_layer_1_neurons = 8
hidden_layer_2_neurons = 5
output_neurons = 1

```

```python
# X positions represent the four layers
x_positions = [0, 1, 2, 3]

# Y positions distribute the neurons vertically
input_y = np.linspace(0, 7, input_neurons)
hidden1_y = np.linspace(0, 7, hidden_layer_1_neurons)
hidden2_y = np.linspace(0, 7, hidden_layer_2_neurons)
output_y = np.array([3.5])


```

```python
#Create the figure

plt.figure(figsize=(20, 14))
```

```python

#connections between Input and Hidden Layer 1
for y1 in input_y:
    for y2 in hidden1_y:

        # Draw a line between two neurons
        plt.plot(
            [x_positions[0], x_positions[1]],
            [y1, y2],
            color="gray",
            linewidth=0.5,
            alpha=0.5
        )



#connections between Hidden Layer 1 and 2


for y1 in hidden1_y:
    for y2 in hidden2_y:

        plt.plot(
            [x_positions[1], x_positions[2]],
            [y1, y2],
            color="gray",
            linewidth=0.5,
            alpha=0.5
        )



#connections between Hidden Layer 2 and Output


for y1 in hidden2_y:
    for y2 in output_y:

        plt.plot(
            [x_positions[2], x_positions[3]],
            [y1, y2],
            color="gray",
            linewidth=0.5,
            alpha=0.5
        )



#Draw neurons in the Input Layer

plt.scatter(
    [x_positions[0]] * input_neurons,
    input_y,
    s=1000,
    color="blue",
    edgecolor="black",
    zorder=3
)

#Draw neurons in Hidden Layer 1
plt.scatter(
    [x_positions[1]] * hidden_layer_1_neurons,
    hidden1_y,
    s=1000,
    color="lightgreen",
    edgecolor="black",
    zorder=3
)
#Draw neurons in Hidden Layer 2

plt.scatter(
    [x_positions[2]] * hidden_layer_2_neurons,
    hidden2_y,
    s=1000,
    color="pink",
    edgecolor="black",
    zorder=3
)

#Draw the Output Neuron


plt.scatter(
    [x_positions[3]],
    output_y,
    s=1000,
    color="red",
    edgecolor="black",
    zorder=3
)


# Label the input neurons

for i, feature in enumerate(input_features):

    plt.text(
        x_positions[0] - 0.15,
        input_y[i],
        feature,
        fontsize=10,
        ha="right",
        va="center"
    )
#Label each Neural Network layer


plt.text(
    x_positions[0],
    8.2,
    "Input Layer\n6 Features",
    fontsize=13,
    fontweight="bold",
    ha="center"
)

plt.text(
    x_positions[1],
    8.2,
    "Hidden Layer 1\n8 Neurons",
    fontsize=13,
    fontweight="bold",
    ha="center"
)

plt.text(
    x_positions[2],
    8.2,
    "Hidden Layer 2\n5 Neurons",
    fontsize=13,
    fontweight="bold",
    ha="center"
)

plt.text(
    x_positions[3],
    8.2,
    "Output Layer\n1 Neuron",
    fontsize=13,
    fontweight="bold",
    ha="center"
)



#output neuron
plt.text(
    x_positions[3] + 0.15,
    output_y[0],
    "Fraud\n0 = Genuine\n1 = Fraud",
    fontsize=11,
    ha="left",
    va="center"
)


#Hide the X and Y axes
plt.axis("off")

#Display the final architecture
plt.show()
```
