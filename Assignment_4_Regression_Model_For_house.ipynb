```python
#required libraries
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
#Machine Learning library
from sklearn.linear_model import LinearRegression
#Library for 3D plotting
from mpl_toolkits.mplot3d import Axes3D
```

```python

# Area (sq ft)
# Number of Bedrooms
# House Price (in Lakhs)

data = {
    'Area': [1000, 1200, 1500, 1800, 2000, 2200, 2500, 2700, 3000],
    'Bedrooms': [2, 2, 3, 3, 4, 4, 4, 5, 5],
    'Price': [30, 35, 45, 50, 60, 65, 75, 80, 90]
}

# Create DataFrame
df = pd.DataFrame(data)

```

```python

#Define Independent Variables

X = df[['Area', 'Bedrooms']]
y = df['Price']

```

```python

#Train Linear Regression Model

model = LinearRegression()
model.fit(X, y)

```

```python

#Create Grid for Regression


area_range = np.linspace(df['Area'].min(),
                         df['Area'].max(), 20)

bedroom_range = np.linspace(df['Bedrooms'].min(),
                            df['Bedrooms'].max(), 20)

A, B = np.meshgrid(area_range, bedroom_range)
```

```python

# Predict house prices
predicted_price = model.predict(
    np.c_[A.ravel(), B.ravel()]
)

#Convert predictions into matrix form
P = predicted_price.reshape(A.shape)

```

```python

#Create 3D Plot

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# actual data points
ax.scatter(
    df['Area'],
    df['Bedrooms'],
    df['Price'],
    color='black',
    s=80,
    label='House Prices'
)

#Plot regression surface
ax.plot_surface(
    A,
    B,
    P,
    cmap='pink',
    alpha=0.6
)


ax.set_xlabel('Area in sq ft')
ax.set_ylabel('Bedrooms')
ax.set_zlabel(' Price (L)')

ax.set_title('3D Visualization')

# Show legend
ax.legend()

# Display plot
plt.show()
```
